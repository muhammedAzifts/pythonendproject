```python
# 1  importing dataset

df = pd.read_csv("/content/social_media_engagement_5000.csv")


# checking the datatypes
df.info()

# no Convertion needed
```

```python
#Convert date columns to datetime

df["posted_at"] = pd.to_datetime(df["posted_at"], dayfirst=True)

# Basic Numpy Opertion
import numpy as np

likes = np.array(df["likes"])
comments = np.array(df["comments"])
shares = np.array(df["shares"])

print("First 5 Likes:", likes[:5])
print("Total Likes:", np.sum(likes))
print("Average Likes:", np.mean(likes))
print("Maximum Likes:", np.max(likes))
print("Minimum Likes:", np.min(likes))

total_engagement = likes + comments + shares

print("First 5 Total Engagement:", total_engagement[:5])
print("Highest Total Engagement:", np.max(total_engagement))
```

```python
# 2 Data Cleaning
# Finding missing values
df.isnull().sum()


```

```python
# Converting missing value in Age to median
df["age"] = df["age"].fillna(df["age"].median())

 #Converting missing value in gender to "unknown"
df["gender"] = df["gender"].fillna("Unknown")
```

```python
#  converting columns likes comments shares to 0
columns = ["likes", "comments", "shares"]

df[columns] = df[columns].fillna(0)
```

```python
#Finf=ding and Removing Duplicate rows
df[df.duplicated()]

```

```python
df = df.drop_duplicates()
```

```python
# converting age and like to int from float, engagement rate to float
df["likes"] = df["likes"].astype(int)
df["age"] = df["age"].astype(int)
df["engagement_rate"] = df["engagement_rate"].astype(float)
```

```python
# Converting female male to m & f
df["gender"] = df["gender"].replace({
    "m": "male",
    "f": "female"
})
```

```python
print("Likes:", df["likes"].min(), df["likes"].max())
print("Comments:", df["comments"].min(), df["comments"].max())
print("Shares:", df["shares"].min(), df["shares"].max())
```

```python
print("Negative likes:", (df["likes"] < 0).sum())
print("Negative comments:", (df["comments"] < 0).sum())
print("Negative shares:", (df["shares"] < 0).sum())
```

```python
# Feature Cleaning
# Extract hashtag count
df["hashtag_count"] = df["hashtags"].fillna("").apply(
    lambda x: len(x.split())
)

df[["hashtags", "hashtag_count"]].head()
```

```python
# cleaning sentimental

df["sentiment"].unique()

df["sentiment"] = df["sentiment"].str.strip().str.lower()
```

```python
# 3   Data Exploration using Pandas

# first 5 row
df.head()
# last 5 rows
df.tail()

#number of rows and column
df.shape

# Column name
df.columns
```

```python
# info and data type
df.info()

df.dtypes


#summenry
df.describe()
```

```python
#categorical distributions using value_counts(), unique(), and nunique().
# Value counts
print("Gender Distribution:")
print(df["gender"].value_counts())

print("\nPost Type Distribution:")
print(df["post_type"].value_counts())

print("\nPost Category Distribution:")
print(df["post_category"].value_counts())

print("\nSentiment Distribution:")
print(df["sentiment"].value_counts())

# Unique values
print("\nUnique Gender Values:")
print(df["gender"].unique())

# Number of unique values
print("\nNumber of Unique Values:")
print("Gender:", df["gender"].nunique())
print("Country:", df["country"].nunique())
print("Post Type:", df["post_type"].nunique())
print("Post Category:", df["post_category"].nunique())
print("Device Type:", df["device_type"].nunique())
print("Sentiment:", df["sentiment"].nunique())
```

```python
# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation
correlation_matrix = numeric_df.corr()

# Display correlation matrix
print(correlation_matrix)
```

```python
# avg likes,comment , share in posttypes category
avg_likes = df.groupby("post_type")["likes"].mean()

print(avg_likes)

avg_comments = df.groupby("post_type")["comments"].mean()

print("\n", avg_comments)


avg_shares = df.groupby("post_type")["shares"].mean()

print(avg_shares)
```

```python
# 4 Data Wrangling
# engagement Score
df["engagement_score"] = (
    df["likes"] +
    df["comments"] +
    df["shares"]
)

# hashtag Count
df["hashtag_count"] = df["hashtags"].fillna("").apply(
    lambda x: len(x.split())
)

# log-transformed metrics (optional)
df["log_likes"] = np.log1p(df["likes"])
df["log_comments"] = np.log1p(df["comments"])
df["log_shares"] = np.log1p(df["shares"])

# display new fields
print(df[[
    "engagement_score",
    "hashtag_count",
    "log_likes",
    "log_comments",
    "log_shares"
]].head())
```

```python
#Summary by post_type
post_type_summary = df.groupby("post_type").agg({
    "likes": "mean",
    "comments": "mean",
    "shares": "mean",
    "impression_count": "mean",
    "engagement_score": "mean"
})

print(post_type_summary)

#Summary by country

country_summary = df.groupby("country").agg({
    "likes": "mean",
    "comments": "mean",
    "shares": "mean",
    "impression_count": "mean",
    "engagement_score": "mean"
})

print(country_summary)

# Summary by sentiment
sentiment_summary = df.groupby("sentiment").agg({
    "likes": "mean",
    "comments": "mean",
    "shares": "mean",
    "impression_count": "mean",
    "engagement_score": "mean"
})

print(sentiment_summary)
```

```python
# 5  Statistical Analysis
# Mean, median, mode
stats = pd.DataFrame({
    "Mean": df[columns].mean(),
    "Median": df[columns].median(),
    "Mode": df[columns].mode().iloc[0]
})

print(stats)
```

```python
# Standard deviation, variance
stats = pd.DataFrame({
    "Standard Deviation": df[columns].std(),
    "Variance": df[columns].var()
})

print(stats)
```

```python
# Percentage
columns = [
    "likes",
    "comments",
    "shares",
    "watch_time_sec",
    "engagement_rate",
    "follower_count"
]

percentiles = df[columns].quantile([0.25, 0.50, 0.75, 0.90])

print(percentiles)
```

```python
#  Which post types have highest engagement?
post_type_engagement = df.groupby("post_type")["engagement_score"].mean().sort_values(ascending=False)

print(post_type_engagement)

#  image have more engagement
```

```python
category_engagement = df.groupby("post_category")["engagement_score"].mean().sort_values(ascending=False)

print(category_engagement)

# music is the best performance content category
```

```python
country_engagement = (
    df.groupby("country")["engagement_rate"]
    .mean()
    .sort_values(ascending=False)
)

print(country_engagement)

# brazil have higest avg engagment rate
```

```python
#

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 25, 35, 45, 100],
    labels=["Under 18", "18-25", "26-35", "36-45", "46+"]
)

age_engagement = df.groupby(
    "age_group",
    observed=True
)["engagement_score"].mean()

print(age_engagement)

#  Under 18 age group has the highest average engagement score (12,634.04),
```

```python
verified_engagement = df.groupby("is_verified")["engagement_score"].mean()

print(verified_engagement)
#  False    12256.345141
# True     12019.985507
```

```python
df["hour"] = df["posted_at"].dt.hour

# Average impressions by hour
hourly_impressions = (
    df.groupby("hour")["impression_count"]
    .mean()
    .sort_values(ascending=False)
)

print(hourly_impressions)

# 12:00 AM (midnight) has an average of 50,013.73 impressions
```

```python
device_watch_time = (
    df.groupby("device_type")["watch_time_sec"]
    .mean()
    .sort_values(ascending=False)
)

print(device_watch_time)

# Mobile users have the highest average watch time at 4,087.83 seconds, followed by tablet users at 3,979.74 seconds and desktop users at 3,974.79 seconds.
```

```python
sentiment_performance = (
    df.groupby("sentiment")["engagement_score"]
    .mean()
    .sort_values(ascending=False)
)

print(sentiment_performance)

# Negative sentiment has the highest average engagement score
```

```python
sentiment_behavior = df[df["sentiment"].isin(["negative", "neutral"])].groupby("sentiment").agg(
    avg_likes=("likes", "mean"),
    avg_comments=("comments", "mean"),
    avg_shares=("shares", "mean"),
    avg_watch_time=("watch_time_sec", "mean"),
    avg_impressions=("impression_count", "mean"),
    avg_engagement_rate=("engagement_rate", "mean")
)

print(sentiment_behavior)


# Negative sentiment posts generally show slightly higher engagement than neutral posts.
```
